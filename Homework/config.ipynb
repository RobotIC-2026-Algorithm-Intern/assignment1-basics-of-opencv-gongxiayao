```python
config = {
  "colors": {
    "blue": [[92, 0, 137], [125, 120, 255]],
    "purple": [[146, 24, 149], [180, 90, 255]]
  },
  "roi": [220, 200, 190, 55],
  "video": "../res/output1.avi"
}
```

```python
import cv2
import numpy as np
import json
import os


class BallDetector:
    def __init__(self):
        self.colors = config["colors"]
        self.roi = config["roi"]
        self.video_path = config["video"]
    
    def detect_color(self, roi):
        if roi.size == 0:
            return "no_roi"
        
        hsv = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)
        max_pixels = 0
        result = "no_ball"
        
        for color_name, (lower, upper) in self.colors.items():
            mask = cv2.inRange(hsv, np.array(lower), np.array(upper))
            count = np.sum(mask > 0)
            
            if count > max_pixels and count > 50:
                max_pixels = count
                result = color_name
        
        return result
    
    def run_detection(self):
        cap = cv2.VideoCapture(self.video_path)
        x, y, w, h = self.roi
        
        while cap.isOpened():
            ret, frame = cap.read()
            if not ret:
                break
            
            roi = frame[y:y+h, x:x+w]
            color = self.detect_color(roi)
            
            # 显示结果
            cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
            cv2.putText(frame, f"Color: {color}", (x, y-10), 
                       cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)
            
            cv2.imshow("Detection", frame)
            if cv2.waitKey(25) & 0xFF == ord('q'):
                break
        
        cap.release()
        cv2.destroyAllWindows()

# 运行检测
if __name__ == "__main__":
    detector = BallDetector()
    detector.run_detection()
```
